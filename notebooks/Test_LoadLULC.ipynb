```python
import pystac
from odc.stac import load

url = "https://s3.us-west-2.amazonaws.com/dep-public-staging/dep_ls_lulc/0-2-0/026/059/2000/dep_ls_lulc_026_059_2000.stac-item.json"
item = pystac.Item.from_file(url)

ds = load([item], chunks={}, bands=["classification"])
ds
```

```python
import re

from matplotlib.colors import ListedColormap

from ldn.typology import colors

# from ldn.utils import LULC_NODATA
LULC_NODATA = 255

# One colour per class code 1-7, in order; vmin/vmax centre each code on its colour.
codes = sorted(c for c in colors if c != LULC_NODATA)
cmap = ListedColormap([tuple(int(v) / 255 for v in re.findall(r"\d+", colors[c])) for c in codes])
classes = ds["classification"].squeeze("time")
classes.where(classes != LULC_NODATA).odc.explore(cmap=cmap, vmin=0.5, vmax=len(codes) + 0.5)
```
